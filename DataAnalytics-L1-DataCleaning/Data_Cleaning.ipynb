```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[(df['Age'] < lower_limit) | (df['Age'] > upper_limit)]

print("Number of Age outliers:", len(outliers))
```

```python
outliers[['PassengerId', 'Name', 'Age']]
```

### Age Outlier Analysis

The IQR method was used to detect outliers in the Age column. The calculated lower limit was -6.6875 and the upper limit was 64.8125. A total of 11 potential outliers were identified.

After inspecting these records, the ages were found to range from 65 to 80 years. These are realistic passenger ages and not data-entry errors. Therefore, all 11 records were retained without modification.

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_limit) | (df['Fare'] > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Fare outliers:", len(outliers))
```

```python
outliers[['PassengerId', 'Pclass', 'Fare']].head(15)
```

### Fare Outlier Analysis

The IQR method was applied to the Fare column. The calculated lower limit was -26.724 and the upper limit was 65.6344. A total of 116 potential outliers were identified.

After inspecting the records, many passengers with high fares were found to belong to First Class. These values are realistic ticket prices and do not necessarily indicate errors. Therefore, all 116 potential outliers were retained without modification.

## Data Cleaning Decisions

1. **Age:** Replace missing values with the median because age data may contain outliers.
2. **Cabin:** Drop the column because approximately 77% of its values are missing.
3. **Embarked:** Replace missing values with the mode because only two values are missing.
4. **Duplicates:** No duplicate rows were found, so no removal is required.
5. **Age outliers:** Retain all 11 outliers because they represent realistic passenger ages.
6. **Fare outliers:** Retain all 116 outliers because they represent plausible ticket prices.
7. **Standardisation:** Existing Sex and Embarked categories are consistent; no formatting changes are needed.
8. **PassengerId:** Convert the column to string because it is an identifier, not a numerical measurement.
9. **Fare:** Retain its float data type because it represents monetary values.

```python
df_cleaned = df.copy()

df_cleaned.shape
```

```python
age_median = df_cleaned['Age'].median()

print("Median Age:", age_median)
```

```python
df_cleaned['Age'] = df_cleaned['Age'].fillna(age_median)

df_cleaned['Age'].isnull().sum()
```

```python
df_cleaned = df_cleaned.drop(columns=['Cabin'])

df_cleaned.shape
```

```python
embarked_mode = df_cleaned['Embarked'].mode()[0]

print("Mode of Embarked:", embarked_mode)
```

```python
df_cleaned['Embarked'] = df_cleaned['Embarked'].fillna(embarked_mode)

df_cleaned['Embarked'].isnull().sum()
```

```python
df_cleaned['PassengerId'] = df_cleaned['PassengerId'].astype(str)

df_cleaned['PassengerId'].dtype
```

### Before vs After Data Quality Analysis

The original Titanic dataset contained 891 rows and 12 columns, with 866 missing values across the Age, Cabin and Embarked columns. After cleaning, the dataset contains 891 rows and 11 columns with zero missing values.

Missing Age values were replaced with the median, missing Embarked values were filled using the mode, and the Cabin column was removed due to approximately 77.1% missing data. No duplicate rows were found before or after cleaning. Genuine Age and Fare outliers were retained after inspection.

The PassengerId column was converted to string because it represents an identifier. All passenger records were preserved, resulting in a clean dataset ready for further analysis.
